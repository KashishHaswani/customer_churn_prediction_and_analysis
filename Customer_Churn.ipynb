**Graphs based on EDA to explain churning in more detail**

```python
churn_counts = df['Churn'].value_counts()
plt.figure(figsize=(5,4))
plt.bar(churn_counts.index, churn_counts.values)

plt.title('Customer Churn Distribution')
plt.xlabel('Churn')
plt.ylabel('Number of Customers')

plt.show()
```

**It visually communicates:**

Most customers stayed.
Around 26.5% churned.
The dataset is imbalanced toward No.

```python
tenure_churn=pd.crosstab(
    df['tenure_group'],
    df['Churn'],
    normalize='index'
)*100

tenure_churn.plot(
    kind='bar',
    stacked=True,
    figsize=(7,5)
)

plt.title('Churn Rate by Customer Tenure')
plt.xlabel('tenure Group')
plt.ylabel('percentage by Customers')
plt.legend(title='Churn')

plt.xticks(rotation=0)
plt.show()
```

**It visually communicates:**

0–12 months    → 47% churn

vs

13–24 months   → 29%

25–48 months   → 20%

49–72 months   → 10%

---




**Key EDA Findings**

1. Customers with shorter tenure show substantially higher churn rates.
   Customers with 0–12 months of tenure have a 47% churn rate,
   compared with 10% among customers with 49–72 months of tenure.

2. Contract type is strongly associated with churn.
   Month-to-month customers have a 42.71% churn rate, compared
   with 11.27% for one-year contracts and 2.83% for two-year contracts.

3. Fiber-optic customers have a higher churn rate (41.89%)
   than DSL customers (18.96%) and customers without internet
   service (7.40%).

4. Electronic-check customers have a churn rate of 45.29%,
   substantially higher than customers using automatic payment methods.

5. Churned customers have higher average monthly charges
   (74.44) than retained customers (61.27), while their average
   total charges are lower because they tend to have shorter tenure.
